```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

df = pd.read_csv("../data/processed/amostra_rotulada.csv")

print(df.head())

# -------------------------
# Distribuição das classes
# -------------------------

contagem = df["classe"].value_counts()

print(contagem)

plt.figure(figsize=(8,5))
contagem.plot(kind="bar")
plt.title("Distribuição das Classes")
plt.xlabel("Classe")
plt.ylabel("Quantidade")
plt.tight_layout()
plt.savefig("../docs/distribuicao_classes.png")
plt.show()

# -------------------------
# Comprimento dos textos
# -------------------------

df["n_palavras"] = df["texto"].str.split().apply(len)

print(df["n_palavras"].describe())

plt.figure(figsize=(8,5))
df["n_palavras"].hist(bins=20)
plt.title("Comprimento dos Textos")
plt.xlabel("Número de palavras")
plt.ylabel("Frequência")
plt.show()

# -------------------------
# Termos mais frequentes
# -------------------------

palavras = " ".join(df["texto"]).lower().split()

frequentes = Counter(palavras)

print("\n20 termos mais frequentes:\n")
print(frequentes.most_common(20))
```
